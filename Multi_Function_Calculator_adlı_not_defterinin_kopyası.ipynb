```python
import sympy
from sympy import symbols, Eq, solve, sqrt
import math

def oran_coz():
    """Oranları çözer (a/b = c/d)."""
    print("\n--- Oran Çözücü ---")
    print("a/b = c/d şeklinde bir oran girin. Bilinmeyen için 'x' kullanın.")
    try:
        a = input("a: ")
        b = input("b: ")
        c = input("c: ")
        d = input("d: ")

        x_sembol = symbols('x')

        # Girdileri sembolik ifadeye dönüştür
        if a.lower() == 'x': a = x_sembol
        if b.lower() == 'x': b = x_sembol
        if c.lower() == 'x': c = x_sembol
        if d.lower() == 'x': d = x_sembol

        # Eğer sayısal değerlerse float'a dönüştür, değilse sembol olarak bırak
        a = float(a) if isinstance(a, str) and a.replace('.', '', 1).isdigit() else a
        b = float(b) if isinstance(b, str) and b.replace('.', '', 1).isdigit() else b
        c = float(c) if isinstance(c, str) and c.replace('.', '', 1).isdigit() else c
        d = float(d) if isinstance(d, str) and d.replace('.', '', 1).isdigit() else d


        if x_sembol not in [a, b, c, d]:
            print("Hata: Oranda 'x' bilinmeyeni bulunamadı.")
            return

        # Denklemi oluştur ve çöz
        # Çapraz çarpım yöntemi: a*d = b*c
        denklem = Eq(a * d, b * c)
        cozum = solve(denklem, x_sembol)

        if cozum:
            print(f"Çözüm: x = {cozum[0]}")
        else:
            print("Bu oran için bir çözüm bulunamadı.")
    except Exception as e:
        print(f"Hata oluştu: {e}. Lütfen geçerli sayılar veya 'x' girin.")

def denklem_x_coz():
    """Doğrusal denklemlerde x'i çözer."""
    print("\n--- Denklemlerde x Çözücü ---")
    print("Doğrusal bir denklem girin (ör: 2*x + 5 = 15).")
    print("Sadece 'x' bilinmeyenini destekler.")
    denklem_str = input("Denklem: ")
    try:
        x = symbols('x')
        # Denklemi sympy formatına dönüştürmek için eval kullanıyoruz.
        # Güvenlik açısından, kullanıcı girdisini doğrudan eval ile çalıştırmak riskli olabilir.
        # Bu basit proje için kabul edilebilir, ancak gerçek uygulamalarda dikkatli olunmalıdır.
        denklem_sag = sympy.sympify(denklem_str)
        # Eğer denklemde '=' yoksa, kullanıcı sadece bir ifade girmiş demektir.
        # Bu durumda ifadeyi 0'a eşitleyelim.
        if '=' in denklem_str:
            sol_taraf, sag_taraf = denklem_str.split('=')
            denklem_sol = sympy.sympify(sol_taraf)
            denklem_sag = sympy.sympify(sag_taraf)
            denklem = Eq(denklem_sol, denklem_sag)
        else:
            denklem = Eq(denklem_sag, 0) # Eğer = yoksa ifadeyi 0'a eşitler

        cozum = solve(denklem, x)
        if cozum:
            print(f"Çözüm: x = {cozum[0]}")
        else:
            print("Bu denklem için bir çözüm bulunamadı veya denklem geçersiz.")
    except Exception as e:
        print(f"Hata oluştu: {e}. Lütfen geçerli bir doğrusal denklem girin.")

def karekok_faktor():
    """Bir sayının karekök çarpanlarını bulur."""
    print("\n--- Karekök Faktörleyici ---")
    try:
        sayi = float(input("Karekökünü faktörlemek istediğiniz sayıyı girin: "))
        if sayi < 0:
            print("Negatif sayıların karekökü reel değildir.")
            return

        if sayi == 0:
            print("Karekök(0) = 0")
            return

        # sympy'nin sqrt fonksiyonu sembolik olarak basitleştirme yapabilir
        basitlestirilmis_kok = sqrt(sayi)
        print(f"√{sayi} = {basitlestirilmis_kok}")

        # Daha detaylı faktörleme için (örneğin 2√3 formatı)
        # Mükemmel kare çarpanını bulma
        tam_sayi_kok = int(math.sqrt(sayi))
        en_buyuk_tam_kare_carpan = 1
        for i in range(1, tam_sayi_kok + 1):
            if sayi % (i*i) == 0:
                en_buyuk_tam_kare_carpan = i*i

        if en_buyuk_tam_kare_carpan > 1:
            disari_cikan = int(math.sqrt(en_buyuk_tam_kare_carpan))
            iceride_kalan = int(sayi / en_buyuk_tam_kare_carpan)
            if iceride_kalan == 1:
                print(f"√{sayi} = {disari_cikan}")
            else:
                print(f"√{sayi} = {disari_cikan}√{iceride_kalan}")
        elif math.isqrt(int(sayi))**2 == sayi: # Tam kare ise
             print(f"√{sayi} = {int(math.sqrt(sayi))}")
        else:
            print(f"√{sayi} daha fazla basitleştirilemez.")

    except ValueError:
        print("Geçersiz giriş. Lütfen bir sayı girin.")
    except Exception as e:
        print(f"Hata oluştu: {e}")

def ondalik_donustur():
    """Ondalık sayıları kesirlere ve yüzdelere dönüştürür."""
    print("\n--- Ondalık Sayı Dönüştürücü ---")
    try:
        ondalik = float(input("Bir ondalık sayı girin: "))

        # Yüzdeye dönüştürme
        yuzde = ondalik * 100
        print(f"Yüzde: %{yuzde:.2f}") # Virgülden sonra 2 basamak

        # Kesire dönüştürme (sympy'nin Rational'ı en iyisi)
        kesir = sympy.Rational(ondalik)
        print(f"Kesir: {kesir}")

    except ValueError:
        print("Geçersiz giriş. Lütfen bir ondalık sayı girin.")
    except Exception as e:
        print(f"Hata oluştu: {e}")

def kesir_donustur():
    """Kesirleri ondalık sayılara ve yüzdelere dönüştürür."""
    print("\n--- Kesir Dönüştürücü ---")
    try:
        pay = int(input("Payı girin: "))
        payda = int(input("Paydayı girin: "))

        if payda == 0:
            print("Payda sıfır olamaz.")
            return

        # Ondalığa dönüştürme
        ondalik = pay / payda
        print(f"Ondalık: {ondalik}")

        # Yüzdeye dönüştürme
        yuzde = ondalik * 100
        print(f"Yüzde: %{yuzde:.2f}")

    except ValueError:
        print("Geçersiz giriş. Lütfen tam sayı değerler girin.")
    except Exception as e:
        print(f"Hata oluştu: {e}")

def yuzde_donustur():
    """Yüzdeleri ondalık sayılara ve kesirlere dönüştürür."""
    print("\n--- Yüzde Dönüştürücü ---")
    try:
        yuzde = float(input("Bir yüzde değeri girin (örn: 75 veya 75%): ").replace('%', ''))

        # Ondalığa dönüştürme
        ondalik = yuzde / 100
        print(f"Ondalık: {ondalik}")

        # Kesire dönüştürme
        kesir = sympy.Rational(ondalik)
        print(f"Kesir: {kesir}")

    except ValueError:
        print("Geçersiz giriş. Lütfen bir yüzde değeri girin.")
    except Exception as e:
        print(f"Hata oluştu: {e}")

def main():
    """Ana menü ve hesap makinesi döngüsü."""
    while True:
        print("\n--- Çok Fonksiyonlu Hesap Makinesi ---")
        print("1. Oranları Çöz")
        print("2. Denklemlerde x'i Çöz")
        print("3. Karekök Faktörle")
        print("4. Ondalık Sayıyı Dönüştür (Kesir/Yüzde)")
        print("5. Kesiri Dönüştür (Ondalık/Yüzde)")
        print("6. Yüzdeyi Dönüştür (Ondalık/Kesir)")
        print("7. Çıkış")

        secim = input("Lütfen yapmak istediğiniz işlemi seçin (1-7): ")

        if secim == '1':
            oran_coz()
        elif secim == '2':
            denklem_x_coz()
        elif secim == '3':
            karekok_faktor()
        elif secim == '4':
            ondalik_donustur()
        elif secim == '5':
            kesir_donustur()
        elif secim == '6':
            yuzde_donustur()
        elif secim == '7':
            print("Hesap makinesinden çıkılıyor. Güle güle!")
            break
        else:
            print("Geçersiz seçim. Lütfen 1 ile 7 arasında bir sayı girin.")

if __name__ == "__main__":
    main()
```
